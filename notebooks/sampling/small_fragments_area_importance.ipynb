## Processing

```python
# Reading and filtering data

fs_polygons = gpd.read_file(path_fs_polygons, layer='VS_Amazonia_2022')

# fs_polygons = fs_polygons.loc[fs_polygons['AREA_Km²'] >= 0.05]

fs_polygons = (
    fs_polygons
        .drop(columns=['VALUE', 'CLASSE', 'ANO', 'AREA_ha'])
        .rename(columns={'AREA_Km²': 'area'})
)
```

```python
fs_small = fs_polygons.loc[fs_polygons['area'] < 0.05]
fs_others = fs_polygons.loc[fs_polygons['area'] >= 0.05]
```

```python
n_small = fs_small.shape[0]
area_small = fs_small.area.sum()

n_others = fs_others.shape[0]
area_others = fs_others.area.sum()
```

```python
p_count_small = n_small / (n_small + n_others)
p_area_small = area_small / (area_small + area_others)
```
